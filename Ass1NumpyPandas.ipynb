## 1. Create a 1D NumPy Array:

```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

## 2. Inspection and Properties:

```python
#shape, data type, and the number of elements in the array.Inspect the shape of the temperatures_w1 array.

print("Shape of the array : ",temperatures_w1.shape)
print("Data type of the array : ",temperatures_w1.dtype)
print("Number of array elements :",temperatures_w1.size)
```

## 3. Array Operations:

```python
# Convert the temperatures_w1 from Celsius to Fahrenheit
temp_farenheit=(temperatures_w1*9)/5+32
print("Temperature in Farenheit : ",temp_farenheit)
```

```python
#Find the maximum, minimum, and mean temperatures for the week.
print("Maximum temperature for the week :",np.max(temp_farenheit))
print("Minimum temperature for the week :",np.min(temp_farenheit))
print("Mean temperature for the week :",np.mean(temp_farenheit))
```

## 4. Array Slicing and Indexing:


```python
print(temp_farenheit)
```

```python
#Extract the temperatures for the first three days of the week.
print(temp_farenheit[:3])
```

```python
#Extract the temperatures for the weekend (last two days).
print(temp_farenheit[-2:])
```

```python
#Extract the temperatures for the middle three days of the week.
print(temp_farenheit[2:5])
```

## 5. Create a 2D Array:

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)
```

```python
#Inspect and print the shape, data type, and total number of elements in the temperatures array.

print("Shape of the array : ",temperatures.shape)
print("Data type of the array : ",temperatures.dtype)
print("Number of array elements :",temperatures.size)
```

```python
#Extract the temperatures for each week and the weekends (last two days)for both weeks.
print("Week 1 temperatures :",temperatures[0])
print("Week 2 temperatures :",temperatures[1])
print("Weekend temperatures for both weeks :",temperatures[:,-2:])
```

# Pandas Series:

## 1. Creating Pandas Series:

```python
student_marks=[95,92,89,85,80]
rank_indices=['Rank1','Rank2','Rank3','Rank4','Rank5']
marks=pd.Series(student_marks,rank_indices,name="Marks")
print(marks)
```

## 2. Indexing and Slicing:

```python
#● Use the integer index position to access the mark of the 1st rank student.
print(marks[0])
```

```python
#● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
marks.loc['Rank1':'Rank3']
```

```python
#● Use the iloc accessor to retrieve the mark of the 3rd rank student.
print(marks.iloc[2])
```

```python
#● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
```

```python
print(marks>90)
```

## Manipulating Series:

```python
#Modify the mark of the student with the 1st rank to 100.
marks['Rank1']=100
print(marks)
```

```python
#● Remove the entry corresponding to the last rank from the Series.
marks.drop('Rank5')
```

```python
#● Compute the CGPA by dividing each mark in the Series by 10.
cgpa=marks/10
print(cgpa)
```
